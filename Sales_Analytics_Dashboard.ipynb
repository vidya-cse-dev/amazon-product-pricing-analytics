```python
import pandas as pd

filename = next(iter(uploaded))

df = pd.read_csv(filename)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
# Inspect data types and non-null values
print("\nDataset information:")
df.info()

# Count missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Count duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Display sample records
df.head(5)
```

```python
import pandas as pd
import numpy as np

# Keep the original dataset unchanged
df_clean = df.copy()

# Convert prices and percentage strings into numbers
df_clean["discounted_price"] = pd.to_numeric(
    df_clean["discounted_price"].str.replace(r"[₹,]", "", regex=True),
    errors="coerce"
)

df_clean["actual_price"] = pd.to_numeric(
    df_clean["actual_price"].str.replace(r"[₹,]", "", regex=True),
    errors="coerce"
)

df_clean["discount_percentage"] = pd.to_numeric(
    df_clean["discount_percentage"].str.replace("%", "", regex=False),
    errors="coerce"
)

# Convert ratings and rating counts into numbers
df_clean["rating"] = pd.to_numeric(
    df_clean["rating"], errors="coerce"
)

df_clean["rating_count"] = pd.to_numeric(
    df_clean["rating_count"].str.replace(",", "", regex=False),
    errors="coerce"
)

# Check the results
print("Updated data types:")
print(df_clean[[
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]].dtypes)

print("\nMissing values after conversion:")
print(df_clean[[
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]].isnull().sum())

df_clean.head()
```

```python
print("Dataset shape:", df_clean.shape)

print("\nData types:")
print(df_clean[[
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]].dtypes)

print("\nMissing values:")
print(df_clean[[
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]].isnull().sum())

print("\nSummary statistics:")
display(df_clean[[
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]].describe())
```

```python
# Check for invalid price and discount values
print("Invalid discounted prices:",
      (df_clean["discounted_price"] <= 0).sum())

print("Invalid actual prices:",
      (df_clean["actual_price"] <= 0).sum())

print("Discount percentages outside 0–100:",
      (~df_clean["discount_percentage"].between(0, 100)).sum())

print("Ratings outside 1–5:",
      (~df_clean["rating"].dropna().between(1, 5)).sum())

# Calculate the absolute price reduction
df_clean["price_savings"] = (
    df_clean["actual_price"] - df_clean["discounted_price"]
)

# Check for discounted prices greater than actual prices
print("Products priced above actual price:",
      (df_clean["price_savings"] < 0).sum())

# Display the products with the highest price savings
display(
    df_clean[
        ["product_name", "actual_price",
         "discounted_price", "discount_percentage",
         "price_savings"]
    ]
    .sort_values("price_savings", ascending=False)
    .head(10)
)
```

```python
# Extract the main category from the category hierarchy
df_clean["main_category"] = (
    df_clean["category"].str.split("|").str[0]
)

# Summarize products by main category
category_analysis = (
    df_clean.groupby("main_category")
    .agg(
        product_count=("product_id", "nunique"),
        avg_actual_price=("actual_price", "mean"),
        avg_discounted_price=("discounted_price", "mean"),
        avg_discount_percentage=("discount_percentage", "mean"),
        avg_rating=("rating", "mean"),
        total_listed_price_savings=("price_savings", "sum")
    )
    .round(2)
    .sort_values("product_count", ascending=False)
)

display(category_analysis)
```

Results:

* Electronics has the largest number of listings: 490, with an average actual listed price of ₹10,127.31.



* Computers & Accessories has the highest average discount percentage among the major categories: 54.02%.



* Office Products has the highest average rating at 4.31, but it contains only 31 listings.



* Some categories contain just one or two products, so their averages shouldn't be treated as representative of the whole category.

```python
import matplotlib.pyplot as plt

# Sort categories by average discount
discount_chart = category_analysis.sort_values(
    "avg_discount_percentage",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    discount_chart.index,
    discount_chart["avg_discount_percentage"]
)

plt.title("Average Listed Discount by Product Category")
plt.xlabel("Average Discount (%)")
plt.ylabel("Product Category")
plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()

plt.show()
```

**Key Insight:**

Home Improvement products have the highest average discount **(57.5%)**, while Office Products have the lowest **(12.35%)**. Electronics and Computers & Accessories also offer high discounts. This shows that **discounts vary across product categories**.

**AVERAGE LISTED PRICE BY PRODUCT CATEGORY**

```python
import matplotlib.pyplot as plt

# Sort categories by average actual price
price_chart = category_analysis.sort_values(
    "avg_actual_price",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    price_chart.index,
    price_chart["avg_actual_price"]
)

plt.title("Average Actual Listed Price by Product Category")
plt.xlabel("Average Price (₹)")
plt.ylabel("Product Category")
plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()

plt.show()
```

**Key Insight:**

Electronics have the highest average listed price, while Toys & games have the lowest. This shows that product prices vary widely across categories. The results represent listed prices in this dataset, not actual selling volumes.

**Discount Percentage vs Customer Rating**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.scatter(
    df_clean["discount_percentage"],
    df_clean["rating"],
    alpha=0.5
)

plt.title("Product Discount vs Customer Rating")
plt.xlabel("Discount Percentage (%)")
plt.ylabel("Customer Rating")
plt.yticks([1, 2, 3, 4, 5])
plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

Products have a wide range of discount percentages, from low discounts to very high discounts.
Customer ratings are mostly concentrated between 3 and 5 stars.
Higher discounts do not necessarily mean higher customer ratings.
Products with similar discount percentages can have different customer ratings.

There is no guarantee that a product will receive a higher customer rating simply because it offers a bigger discount. Other factors, such as product quality and customer experience, may also influence ratings.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select the top 10 products by discount
top_discounts = (
    df_clean
    .sort_values("discount_percentage", ascending=False)
    .drop_duplicates(subset="product_id")
    .head(10)
    .copy()
)

# Prepare data for heatmap
heatmap_data = top_discounts.set_index(
    top_discounts["product_name"].str.slice(0, 35)
)[["discount_percentage"]]

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd",
    linewidths=0.5,
    cbar_kws={"label": "Discount (%)"}
)

plt.title("Top 10 Products with Highest Discounts")
plt.xlabel("Discount Percentage")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

**Key Insights**

The highest listed discount is 94%, offered on a Mini USB Type-C adapter.
Fire-Boltt Ninja Call Pro Plus products have discounts of around 91%. Several other products, including charging cables and accessories, have discounts of approximately 90%. The heatmap uses darker red shades for higher discounts and lighter yellow shades for lower discounts.

The selected products offer very high listed discounts, ranging from 90% to 94%. Small electronic accessories and smartwatch listings feature prominently among the top discounted products.

```python
import matplotlib.pyplot as plt
import seaborn as sns


numeric_cols = [
    "actual_price",
    "discounted_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

correlation_matrix = df_clean[numeric_cols].corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Amazon Product Features")
plt.tight_layout()
plt.show()
```

**Key Insights**

* Values near +1 indicate a strong positive relationship.



* Values near −1 indicate a strong negative relationship.



* Values near 0 indicate a weak linear relationship.



* The diagonal will always show 1.00 because each feature is perfectly correlated with itself.
